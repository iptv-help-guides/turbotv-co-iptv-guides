## Why are TurboTV movies not playing on my app?
When movies and series fail to load, the most common culprit is an outdated [Video on Demand](https://turbotv.co/what-is-ott/) (VOD) playlist. Streaming applications require regular synchronization to fetch the latest server paths for media files. If the library has been updated but your device has not synced, the app will attempt to pull from a dead link, resulting in a playback error. To resolve this:
1. Navigate to your application playlist settings.
2. Select the option to update or refresh the VOD content.
3. Wait for the download prompt to finish.
4. Exit the application entirely and relaunch it.
Many modern apps allow you to refresh movies and series independently of live channels. This simple synchronization often restores full access to your favorite on-demand content without requiring further technical adjustments.

## How do player decoder settings affect VOD playback?
If your playlist is fully updated but videos still refuse to launch, your player decoder settings might be the issue. Media players use decoders to translate video files into the images you see on screen. Most apps offer a choice between Hardware (HW) and Software (SW) decoding. Hardware decoding relies on your device built-in processing chips, which is generally more efficient, but it can fail if the specific movie file uses an incompatible format. Software decoding uses the app internal processing, which is highly compatible but requires more processing power. If a movie is stuck on a black screen or buffering indefinitely:
1. Open your app playback settings.
2. Locate the media decoder options.
3. Switch the decoder from HW to SW, or vice versa.
Restart the movie to see if the playback issue is resolved.

## What should I do about unsupported formats and codecs?
Video on Demand libraries contain files encoded in various formats, such as MP4, MKV, or AVI, using different video and audio codecs like H.264, H.265 (HEVC), or AAC. Occasionally, the default media player integrated into your streaming application may lack the necessary software instructions to read a specific codec. When this happens, the movie might play with no sound, display a distorted picture, or simply crash the app. To bypass format restrictions, check if your application allows you to use an external media player. Downloading a robust, widely compatible third-party player from your device official app store and setting it as the default player for VOD content can instantly solve format-related playback errors. These dedicated players contain extensive codec libraries designed to handle almost any media file.

## Can subtitles or multiple audio tracks cause VOD errors?
Movies and series frequently include multiple audio tracks for different languages and embedded subtitle files. While these features enhance the viewing experience, they can sometimes cause playback instability, especially on older or less powerful streaming devices. Large embedded subtitle files, particularly those in image-based formats, can overwhelm the player memory allocation, causing the stream to freeze or crash shortly after starting. Similarly, switching between high-bitrate audio tracks might disrupt the video feed. If a specific movie fails to play smoothly, try disabling the subtitles entirely before starting the stream. If the video plays normally without subtitles, you can attempt to download a lighter, text-based external subtitle file if your app supports it. Ensuring you select the standard stereo audio track rather than a surround sound option can also improve stability.

## How does device storage and cache impact streaming?
Streaming on-demand content requires a certain amount of temporary storage space on your device to buffer the video feed smoothly. As you use your streaming application, it accumulates temporary files known as cache. Over time, a bloated cache can consume your device available storage, leaving insufficient room for the media player to buffer new movies or series. This lack of space often manifests as constant freezing, sudden app closures, or movies that refuse to load entirely. To maintain optimal performance, follow these steps:
1. Navigate to your device system settings.
2. Locate your streaming application in the app list.
3. Select the option to clear the cache.
Avoid clearing the application data, as this will erase your login credentials and playlists. Ensuring your device maintains at least one gigabyte of free internal storage will significantly reduce local playback errors.

## Is my internet speed and stability sufficient for VOD?
While on-demand content can sometimes buffer more forgivingly than live broadcasts, your network connection remains crucial. High-definition (HD) streaming requires a baseline of about 10 Mbps per screen. For Full HD (1080p), you need approximately 25 Mbps, and for 4K movies, a minimum of 40+ Mbps is necessary. However, raw speed is only part of the equation; network stability matters significantly more than peak speed. Minor fluctuations or packet loss can interrupt the connection to the VOD server, causing the movie to stall. To ensure the best possible [connection](https://turbotv.co/streaming-quality-guide/):
1. Hardwire your streaming device to your router using an Ethernet cable.
2. If you must use Wi-Fi, position your device close to the router.
3. Connect to the 5 GHz band to minimize interference from other household electronics.

## When should I contact support for VOD issues?
If you have updated your playlist, cleared your cache, adjusted your decoders, and verified your network stability, the playback issue might originate from the server side. Occasionally, specific movie or series files can become corrupted on the host server, or the server itself might be undergoing maintenance. Before reaching out for assistance, use the check below to run a quick diagnostic on your current network environment. If the diagnostic shows no local errors, it is time to [contact your provider customer service team](https://turbotv.co/contact/). Provide them with the exact title of the movie or series, the season and episode number, and the specific [error code](https://turbotv.co/iptv-error-codes/) you are encountering. This detailed information allows the technical team to locate the exact file and repair or replace it swiftly.

## 🎞️ HLS (.m3u8) Stream Analyzer
Inspects an HLS master playlist: renditions, bitrates, resolutions, codecs, segment length, live or VOD — and the speed each quality needs.

By default the notebook analyses **Apple’s public HLS test stream** (an official developer example). Replace the URL with a stream you are entitled to use.

> Speed measurements run from Google’s Colab servers, not from your home.

```python
#@title HLS stream URL { display-mode: "form" }
STREAM_URL = "https://devstreaming-cdn.apple.com/videos/streaming/examples/img_bipbop_adv_example_fmp4/master.m3u8"  #@param {type:"string"}

import re, time, requests
from urllib.parse import urljoin
import pandas as pd

UA = {"User-Agent": "Mozilla/5.0 (HLS analyzer notebook)"}
master = requests.get(STREAM_URL.strip(), timeout=30, headers=UA)
master.raise_for_status()
text = master.text
if not text.lstrip().startswith("#EXTM3U"):
    raise ValueError("This is not an HLS playlist (#EXTM3U missing).")

ATTR = re.compile(r'([A-Z0-9-]+)=("[^"]*"|[^,]*)')
def attrs(line):
    return {k: v.strip('"') for k, v in ATTR.findall(line.split(":", 1)[1])}

variants, media = [], []
lines = text.splitlines()
for i, line in enumerate(lines):
    if line.startswith("#EXT-X-STREAM-INF"):
        a = attrs(line)
        uri = next((l.strip() for l in lines[i + 1:] if l.strip() and not l.startswith("#")), "")
        bw = int(a.get("AVERAGE-BANDWIDTH") or a.get("BANDWIDTH") or 0)
        variants.append({
            "resolution": a.get("RESOLUTION", "audio only"),
            "Mbps": round(bw / 1e6, 2),
            "peak_Mbps": round(int(a.get("BANDWIDTH", 0)) / 1e6, 2),
            "fps": a.get("FRAME-RATE", ""),
            "codecs": a.get("CODECS", ""),
            "uri": urljoin(STREAM_URL, uri),
        })
    elif line.startswith("#EXT-X-MEDIA"):
        a = attrs(line)
        media.append({"type": a.get("TYPE"), "name": a.get("NAME"), "language": a.get("LANGUAGE", ""), "default": a.get("DEFAULT", "")})

if variants:
    v = pd.DataFrame(variants).sort_values("Mbps")
    print(f"Master playlist: {len(v)} renditions, {len(media)} audio/subtitle tracks")
    display(v.drop(columns=["uri"]))
    if media:
        display(pd.DataFrame(media))
else:
    print("Single media playlist (one quality).")
    v = pd.DataFrame([{"resolution": "?", "Mbps": None, "peak_Mbps": None, "fps": "", "codecs": "", "uri": STREAM_URL}])
```

```python
#@title Bitrate per quality (chart)
import matplotlib.pyplot as plt
vv = v.dropna(subset=["Mbps"])
if len(vv):
    ax = vv.plot.bar(x="resolution", y=["Mbps", "peak_Mbps"], figsize=(9, 3.5), color=["#22d3ee", "#b6f23a"], rot=30)
    ax.set_ylabel("Mbps")
    ax.set_title("Average and peak bitrate per rendition")
    plt.tight_layout()
    plt.show()
    best = vv.iloc[-1]
    print(f"For the best quality ({best['resolution']}): plan for at least {max(best['peak_Mbps'] or 0, best['Mbps'] or 0) * 1.5:.1f} Mbps steady per screen (peak bitrate + 50% headroom).")
```

```python
#@title Segments and download speed
target = v.iloc[len(v) // 2]["uri"]  # a mid-quality rendition
pl = requests.get(target, timeout=30, headers=UA).text
durations = [float(x) for x in re.findall(r"#EXTINF:([\d.]+)", pl)]
live = "#EXT-X-ENDLIST" not in pl
seg_uris = [l.strip() for l in pl.splitlines() if l.strip() and not l.startswith("#")]
kind = "live" if live else "on demand (VOD)"
print(f"Type: {kind}")
if durations:
    print(f"Segments: {len(durations)} · average length {sum(durations) / len(durations):.1f} s · total {sum(durations) / 60:.1f} min")
if seg_uris:
    seg = urljoin(target, seg_uris[0])
    MAX_BYTES, MAX_SECONDS = 8_000_000, 20  # keep the test short
    t0, size, complete = time.time(), 0, True
    with requests.get(seg, timeout=15, headers=UA, stream=True) as r:
        for chunk in r.iter_content(65536):
            size += len(chunk)
            if size >= MAX_BYTES or time.time() - t0 > MAX_SECONDS:
                complete = False
                break
    dt = max(time.time() - t0, 0.001)
    print(f"First segment: {size / 1e6:.2f} MB in {dt:.2f} s → {size * 8 / 1e6 / dt:.1f} Mbps (from Colab)")
    seg_len = durations[0] if durations else 0
    if seg_len and complete:
        print(f"Actual segment bitrate: {size * 8 / 1e6 / seg_len:.2f} Mbps")
```

## Key points
- For live TV the **peak bitrate** matters more than the average — keep headroom.
- **Short segments** (2–6 s) start faster and adapt to the network more quickly.
- Several renditions = **adaptive** playback: the player lowers the quality instead of stalling when the network slows down.

## FAQ
**Why does a movie play audio but show a black screen?**
A black screen with working audio usually indicates a video codec incompatibility. Switching your app media decoder from Hardware (HW) to Software (SW), or using a different external media player, will typically resolve this issue.

**How often should I refresh my VOD playlist?**
It is good practice to refresh your Video on Demand playlist every few days or whenever you notice missing titles. Regular updates ensure your app has the correct server paths for the latest movies and series.

**Does clearing the app cache delete my saved favorites?**
No, clearing the temporary cache only removes temporary playback files and will not delete your favorites, watch history, or login credentials. However, be careful not to select the clear data option, as that will reset the entire application.
